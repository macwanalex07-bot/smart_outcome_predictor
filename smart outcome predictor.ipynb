# Part A: Conceptual Foundation (Theory)

### 1. What are Ensemble Learning Methods and why are they effective?

Ensemble Learning is a machine learning technique that combines multiple models to make better predictions. It improves accuracy and reduces errors compared to using a single model.

### 2. Explain the difference between Bagging, Boosting, and Voting.

- Bagging: Trains multiple models independently on different samples of data and combines their predictions. Example: Random Forest.
- Boosting: Trains models one by one, where each new model focuses on correcting previous errors. Example: AdaBoost, XGBoost.
- Voting: Combines predictions from multiple models and selects the final output using majority voting or average probability. Example: Voting Classifier.
  
### 3. What is Bias-Variance Trade-off, and how do ensemble methods address it?

Bias is the error caused by a model being too simple. Variance is the error caused by a model being too sensitive to training data. Ensemble methods combine multiple models to balance bias and variance and improve accuracy.

### 4. Explain Voting Classifier and Stacking Ensemble.

- Voting Classifier: Combines predictions from multiple models and gives the final result using majority voting (Hard Voting) or average probabilities (Soft Voting).
- Stacking Ensemble: Combines multiple models and uses another model called a Meta-model to make the final prediction.

### 5. Compare AdaBoost, Gradient Boosting, LightGBM, and XGBoost conceptually.

- AdaBoost: Focuses more on incorrectly predicted data points by giving them higher weights.
- Gradient Boosting: Builds models one by one, where each new model corrects previous errors using gradients.
- LightGBM: A fast and memory-efficient boosting algorithm that uses leaf-wise tree growth.
- XGBoost: An optimized boosting algorithm that uses regularization to reduce overfitting and improve performance.

All four are boosting algorithms, but they differ in training speed, tree-building methods, and optimization techniques.

```python
from sklearn.impute import SimpleImputer
si = SimpleImputer(strategy="median")

df[["time_spent_hours", "avg_quiz_score", "attendance_rate"]] = si.fit_transform(df[["time_spent_hours", "avg_quiz_score", "attendance_rate"]])
```

```python
df["course_start_date"] = pd.to_datetime(df["course_start_date"])
```

```python
df.drop(columns=["student_id", "course_start_date"], inplace=True)
```

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
df["course_level"] = df["course_level"].map({
    "Beginner":1,
    "Intermediate":2,
    "Advanced":3
})

oe = OneHotEncoder(handle_unknown="ignore", sparse_output=False)

encoded = oe.fit_transform(df[['country_region', 'device_type',
                               'education_background', 'course_category']])

encoded_df = pd.DataFrame(encoded, columns=oe.get_feature_names_out())

df = pd.concat([df.drop(columns=['country_region', 'device_type',
                               'education_background', 'course_category']), encoded_df],axis=1)
```

```python
X = df.drop(["completion_status", "final_score"], axis=1)
y_class = df["completion_status"]
y_reg = df["final_score"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y_class, test_size=0.2, random_state=32, stratify=y_class
)
```

```python
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
   X, y_reg, test_size=0.2, random_state=32
)
```

```python
print("Classification shape:", X_train.shape, X_test.shape)
print("Regression shape:", X_train_reg.shape, X_test_reg.shape)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X_train_scale = scaler.fit_transform(X_train)
X_test_scale = scaler.transform(X_test)

X_train_reg_scale = scaler.fit_transform(X_train_reg)
X_test_reg_scale = scaler.transform(X_test_reg)

```

# Part C: Bagging (Bootstrap Aggregating)

```python
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix, recall_score
print("Accuracy_score:", accuracy_score(y_test, y_pred_clf_bag))
print("f1_score:", f1_score(y_test, y_pred_clf_bag))
print("Classification report:\n", classification_report(y_test, y_pred_clf_bag))
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred_clf_bag))
print("recall_score:\n", recall_score(y_test, y_pred_clf_bag))
```

# Part E: Voting & Stacking Ensembles

```python
from sklearn.ensemble import VotingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression

voting_clf = VotingClassifier(
    estimators=[
        ('dt', DecisionTreeClassifier(random_state=32)),
        ('nb', GaussianNB()),
        ('knn', KNeighborsClassifier(n_neighbors=5)),
        ('lr', LogisticRegression(max_iter=1000))
    ],
    voting='hard'
)

voting_clf.fit(X_train_scale, y_train)

y_pred_voting = voting_clf.predict(X_test_scale)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Accuracy:", accuracy_score(y_test, y_pred_voting))
print("\nClassification Report:\n", classification_report(y_test, y_pred_voting))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_voting))
```

```python
soft_voting = VotingClassifier(
    estimators=[
        ('dt', DecisionTreeClassifier(random_state=32)),
        ('nb', GaussianNB()),
        ('knn', KNeighborsClassifier(n_neighbors=5)),
        ('lr', LogisticRegression(max_iter=1000))
    ],
    voting='soft'
)

soft_voting.fit(X_train_scale, y_train)

y_pred_soft = soft_voting.predict(X_test_scale)
```

```python
print("Hard Voting Accuracy:", accuracy_score(y_test, y_pred_voting))
print("Soft Voting Accuracy:", accuracy_score(y_test, y_pred_soft))
```

```python
from sklearn.ensemble import StackingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.linear_model import LogisticRegression

stack_clf = StackingClassifier(
    estimators=[
        ('dt', DecisionTreeClassifier(random_state=32)),
        ('nb', GaussianNB()),
        ('knn', KNeighborsClassifier(n_neighbors=5))
    ],
    final_estimator=LogisticRegression(),
    cv=5
)

stack_clf.fit(X_train_scale, y_train)

y_pred_stack = stack_clf.predict(X_test_scale)
```

```python

print("Accuracy:", accuracy_score(y_test, y_pred_stack))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_stack))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_stack))
```
